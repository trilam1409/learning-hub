```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "engine_size": [100, 120, 150, 170, 200, 110, 160],
    "price": [9000, 11000, 14000, 17000, 22000, 9800, 15300],
    "body_style": ["sedan", "sedan", "hatchback", "suv", "suv", "hatchback", "sedan"],
})
print(df.describe())
print("Median:", df["price"].median())
print("Skew:", df["price"].skew())
```

```python
df["price"].plot.box()
plt.title("Box plot: price")
plt.show()
df.plot.scatter(x="engine_size", y="price")
plt.show()
```

```python
print(df.corr(numeric_only=True))
print(df.groupby("body_style")["price"].agg(["mean","count"]))
```

```python
from scipy.stats import f_oneway
groups = [group["price"].values for _, group in df.groupby("body_style")]
print(f_oneway(*groups))
```
