Thực hành dữ liệu thiếu, Min-Max, Z-score, pd.cut, one-hot. Các giá trị đều là ví dụ tự tạo.

Chạy các phần theo thứ tự. Hãy thử trong ô tự làm rồi so sánh với ô lời giải bên dưới. Bạn có thể bỏ qua ô tự làm nếu muốn xem ví dụ trước.

Chạy ô chuẩn bị ngay bên dưới. Dùng biến mới trong ô tự làm để giữ nguyên bộ dữ liệu cho lời giải mẫu.

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "price": ["10000", "?", "15000", "20000"],
    "horsepower": [100, np.nan, 130, 160],
    "fuel_type": ["gas", "diesel", "gas", "diesel"],
})
```

## Thử 1

Đổi dấu ? thành giá trị thiếu và chuyển cột price thành số. Dự đoán hàng nào bị thiếu giá.

Gợi ý — Dùng replace rồi to_numeric.

```python
# Viết câu trả lời ở đây.
```

**Đối chiếu 1** — Hàng chỉ số 1 có giá NaN.

Ô bên dưới là lời giải mẫu và chuẩn bị dữ liệu cho phần tiếp theo.

```python
df = df.replace("?", np.nan)
df["price"] = pd.to_numeric(df["price"])
display(df)
```

## Thử 2

Điền horsepower bị thiếu bằng trung vị và dự đoán giá trị được điền.

Gợi ý — Tính median rồi dùng fillna.

```python
# Viết câu trả lời ở đây.
```

**Đối chiếu 2** — Giá trị được điền là 130.

Ô bên dưới là lời giải mẫu và chuẩn bị dữ liệu cho phần tiếp theo.

```python
df["horsepower_filled"] = df["horsepower"].fillna(df["horsepower"].median())
display(df)
```

## Thử 3

Tính Min-Max và Z-score. Dự đoán cả hai kết quả của giá trị 60.

Gợi ý — 60 là trung bình và cũng nằm giữa 20 và 100.

```python
# Viết câu trả lời ở đây.
```

**Đối chiếu 3** — Min-Max bằng 0.5 và Z-score bằng 0.

Ô bên dưới là lời giải mẫu và chuẩn bị dữ liệu cho phần tiếp theo.

```python
values = pd.Series([20, 40, 60, 80, 100], dtype="float64")
minmax = (values - values.min()) / (values.max() - values.min())
zscore = (values - values.mean()) / values.std(ddof=0)
display(pd.DataFrame({"value": values, "minmax": minmax, "zscore": zscore}))
```

## Thử 4

Chia giá thành ba nhóm và mã hóa loại nhiên liệu.

Gợi ý — Dùng pd.cut rồi pd.get_dummies.

```python
# Viết câu trả lời ở đây.
```

**Đối chiếu 4** — Giá 15000 ở nhóm Vừa, loại nhiên liệu có hai cột chỉ thị.

Ô bên dưới là lời giải mẫu và chuẩn bị dữ liệu cho phần tiếp theo.

```python
df["price_group"] = pd.cut(df["price"], bins=[0, 12000, 18000, 30000], labels=["Thấp", "Vừa", "Cao"])
encoded = pd.get_dummies(df, columns=["fuel_type"], dtype=int)
display(encoded)
```
