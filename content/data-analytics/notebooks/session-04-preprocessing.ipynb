# Buổi 04 – Tiền xử lý và chuẩn hóa
Thực hành dữ liệu thiếu, Min-Max, Z-score, pd.cut, one-hot. Các giá trị đều là ví dụ tự tạo.

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "price": ["10000", "?", "15000", "20000"],
    "horsepower": [100, np.nan, 130, 160],
    "fuel_type": ["gas", "diesel", "gas", "diesel"],
})
df = df.replace("?", np.nan)
df["price"] = pd.to_numeric(df["price"])
display(df)
```

```python
df["horsepower_filled"] = df["horsepower"].fillna(df["horsepower"].median())
display(df)
```

```python
values = pd.Series([20, 40, 60, 80, 100], dtype="float64")
minmax = (values - values.min()) / (values.max() - values.min())
zscore = (values - values.mean()) / values.std(ddof=0)
display(pd.DataFrame({"value": values, "minmax": minmax, "zscore": zscore}))
```

```python
df["price_group"] = pd.cut(df["price"], bins=[0, 12000, 18000, 30000], labels=["Thấp", "Vừa", "Cao"])
encoded = pd.get_dummies(df, columns=["fuel_type"], dtype=int)
display(encoded)
```
