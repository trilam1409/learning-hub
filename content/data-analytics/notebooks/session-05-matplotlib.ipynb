```python
import matplotlib.pyplot as plt
import pandas as pd

years = [2021, 2022, 2023, 2024]
sales = [100, 150, 140, 210]
```

**Đối chiếu 1** — Từ 2022 đến 2023.

Ô bên dưới là lời giải mẫu và chuẩn bị dữ liệu cho phần tiếp theo.

```python
plt.plot(years, sales, marker="o")
plt.title("Doanh số theo năm (ví dụ)")
plt.xlabel("Năm")
plt.ylabel("Số lượng")
plt.grid(alpha=0.3)
plt.show()
```

## Thử 2

Vẽ biểu đồ cột cho cùng dữ liệu. Tìm năm có cột cao nhất.

Gợi ý — Dùng kind="bar".

```python
# Viết câu trả lời ở đây.
```

**Đối chiếu 2** — Năm 2024 với 210.

Ô bên dưới là lời giải mẫu và chuẩn bị dữ liệu cho phần tiếp theo.

```python
df = pd.DataFrame({"year": years, "sales": sales})
df.plot(x="year", y="sales", kind="bar", legend=False)
plt.title("So sánh doanh số theo năm")
plt.show()
```
