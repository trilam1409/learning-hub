# Buổi 05 – Matplotlib và Pandas plotting
Bộ dữ liệu tự tạo minh họa cách sử dụng pyplot và DataFrame.plot().

```python
import matplotlib.pyplot as plt
import pandas as pd

years = [2021, 2022, 2023, 2024]
sales = [100, 150, 140, 210]
plt.plot(years, sales, marker="o")
plt.title("Doanh số theo năm (ví dụ)")
plt.xlabel("Năm")
plt.ylabel("Số lượng")
plt.grid(alpha=0.3)
plt.show()
```

```python
df = pd.DataFrame({"year": years, "sales": sales})
df.plot(x="year", y="sales", kind="bar", legend=False)
plt.title("So sánh doanh số theo năm")
plt.show()
```
