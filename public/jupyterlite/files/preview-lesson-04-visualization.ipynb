Nguồn slide: Bai04_Cong_cu_TQ.pdf. Nội dung chưa học trên lớp. Dữ liệu dưới đây là giả lập.

```python
import matplotlib.pyplot as plt
import pandas as pd

years = [2020, 2021, 2022, 2023, 2024]
df = pd.DataFrame({"year": years, "A": [20, 25, 23, 31, 35], "B": [10, 14, 19, 21, 26]})
df.plot(x="year", y=["A", "B"], kind="line", marker="o")
plt.title("Line plot – dữ liệu minh họa")
plt.show()
```

```python
df.set_index("year")[["A", "B"]].plot.area(stacked=True)
plt.title("Stacked area – dữ liệu minh họa")
plt.show()
```

```python
df["A"].plot.hist(bins=4)
plt.title("Histogram – biến A")
plt.show()
```

```python
df[["A", "B"]].sum().plot.pie(autopct="%1.1f%%")
plt.title("Pie – tổng nhóm")
plt.ylabel("")
plt.show()
```
