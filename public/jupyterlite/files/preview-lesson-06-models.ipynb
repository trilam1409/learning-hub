Thực hành LinearRegression, polynomial regression, pipeline, MSE và R². Dữ liệu tự tạo.

Chạy các phần theo thứ tự. Hãy thử trong ô tự làm rồi so sánh với ô lời giải bên dưới. Bạn có thể bỏ qua ô tự làm nếu muốn xem ví dụ trước.

Chạy ô chuẩn bị ngay bên dưới. Dùng biến mới trong ô tự làm để giữ nguyên bộ dữ liệu cho lời giải mẫu.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

X = np.arange(1, 11, dtype=float).reshape(-1, 1)
y = 3 * X.ravel() + 5 + np.array([0, 1, 0, -1, 0, 2, -1, 0, 1, -1])
```

## Thử 1

Fit hồi quy tuyến tính và tính MSE, R² trên các điểm ví dụ.

Gợi ý — Dùng predict để tạo giá trị dự đoán.

```python
# Viết câu trả lời ở đây.
```

**Đối chiếu 1** — MSE không âm. Điểm này đo ngay trên dữ liệu fit.

Ô bên dưới là lời giải mẫu và chuẩn bị dữ liệu cho phần tiếp theo.

```python
lr = LinearRegression().fit(X, y)
predicted = lr.predict(X)
print("Intercept:", lr.intercept_, "Slope:", lr.coef_)
print("MSE:", mean_squared_error(y, predicted), "R²:", r2_score(y, predicted))
```

```python
import matplotlib.pyplot as plt
plt.scatter(X.ravel(), y, label="Observed")
plt.plot(X.ravel(), predicted, label="Predicted")
plt.legend()
plt.show()
plt.scatter(predicted, y - predicted)
plt.axhline(0, color="red", linestyle="--")
plt.show()
```

**Đối chiếu 3** — Lỗi fit chỉ cho biết độ khớp trên các điểm đã dùng, chưa đo khả năng dự đoán điểm mới.

Ô bên dưới là lời giải mẫu và chuẩn bị dữ liệu cho phần tiếp theo.

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LinearRegression

pipeline = make_pipeline(StandardScaler(), PolynomialFeatures(degree=2), LinearRegression())
pipeline.fit(X, y)
print(pipeline.predict([[11]]))
```
