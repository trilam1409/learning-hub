# Đọc trước Bài 06 – Phát triển mô hình
Thực hành LinearRegression, polynomial regression, pipeline, MSE và R². Dữ liệu tự tạo.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

X = np.arange(1, 11, dtype=float).reshape(-1, 1)
y = 3 * X.ravel() + 5 + np.array([0, 1, 0, -1, 0, 2, -1, 0, 1, -1])
lr = LinearRegression().fit(X, y)
predicted = lr.predict(X)
print("Intercept:", lr.intercept_, "Slope:", lr.coef_)
print("MSE:", mean_squared_error(y, predicted), "R²:", r2_score(y, predicted))
```

```python
import matplotlib.pyplot as plt
plt.scatter(X.ravel(), y, label="Observed")
plt.plot(X.ravel(), predicted, label="Predicted")
plt.legend()
plt.show()
plt.scatter(predicted, y - predicted)
plt.axhline(0, color="red", linestyle="--")
plt.show()
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LinearRegression

pipeline = make_pipeline(StandardScaler(), PolynomialFeatures(degree=2), LinearRegression())
pipeline.fit(X, y)
print(pipeline.predict([[11]]))
```
