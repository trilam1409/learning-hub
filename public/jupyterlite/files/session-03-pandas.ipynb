# Buổi 03 – Pandas Series và DataFrame
Thực hành với bộ dữ liệu tự tạo, không cần tải CSV.

```python
import pandas as pd

series_a = pd.Series([10, 20], index=["x", "y"])
series_b = pd.Series([1, 2], index=["y", "z"])
print(series_a + series_b)
```

```python
df = pd.DataFrame({
    "make": ["Toyota", "Honda", "Ford"],
    "price": [12000, 15000, 9000],
    "horsepower": [120, 140, 95],
})
display(df.head())
display(df.describe())
print(df.dtypes)
```

```python
df.to_csv("cars-demo.csv", index=False)
again = pd.read_csv("cars-demo.csv")
display(again)
```
