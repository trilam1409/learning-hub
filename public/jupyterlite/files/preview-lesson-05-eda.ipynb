So sánh giá trung bình giữa nhóm bằng ANOVA. Xác định cách đọc p-value.

```python
# Viết câu trả lời ở đây.
```

**Đối chiếu 4** — p-value lớn không đủ để bác bỏ giả thuyết các trung bình bằng nhau.

Ô bên dưới là lời giải mẫu và chuẩn bị dữ liệu cho phần tiếp theo.

```python
from scipy.stats import f_oneway
groups = [group["price"].values for _, group in df.groupby("body_style")]
print(f_oneway(*groups))
```
